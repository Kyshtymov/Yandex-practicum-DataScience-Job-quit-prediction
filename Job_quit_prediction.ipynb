<b> Описание проекта: </b> Удовлетворённость работой напрямую влияет на отток сотрудников. А предсказание оттока — одна из важнейших задач HR-аналитиков. Внезапные увольнения несут в себе риски для компании, особенно если уходит важный сотрудник. Для этих целей HR-аналитикам пригодится машинное обучение, с помощью которого получится быстрее и точнее отвечать на вопросы бизнеса.

<b> Цель исследования: </b> Построить модель, которая сможет предсказать уровень удовлетворённости сотрудника на основе данных заказчика и построить модель, которая сможет на основе данных заказчика предсказать то, что сотрудник уволится из компании.

<b> Ход исследования для двух задач:</b> 

- Подготовка данных: загрузка и изучение общей информации из представленных датасетов
- Предобработка данных: обработка пропущенных значений и дубликатов
- Исследовательский анализ данных: построение и анализ признаков, выявления аномальных значений и выбросов, поиск корреляций между целевым признаком
- Для второй задачи дополнительно составить портрет уволившегося сотрудника, определить наличие корреляции с ухода сотрудника из компании и его удовлетворённостью работой, а также на основе модели из первой задачи предсказать удовлетворённость работой у сотрудников компаний
- Разработка модели машинного обучения: построение лучшей модели при помощи пайплайнов с различными видами гиперпараметров


<b> Общий вывод: </b> Выбор лучших моделей машинного обучения на основе метрик SMAPE и roc_auc для первой и второй задач и рекомендации по удержанию сотрудников в компании.

При помощи данного исследования бизнес сможет понять, что движет сотрудников на увольнение, и вовремя определить тех, кто планирует уйти, чтобы предпринять контрмеры.

```python
#Напишем функции для отрисовки графиков
def draw_quantitative(df, quantity):
    fig, ax=plt.subplots(nrows=1, ncols=1, figsize=(5, 5))
    sns.set_style("whitegrid")
    sns.histplot(x=df[quantity], ax=ax)
    ax.set(ylabel='Количество событий')
    
```

```python
def draw_categorize(df, quantity, value):
    fig, ax=plt.subplots(nrows=1, ncols=2, figsize=(10, 10))
    sns.set_style("whitegrid")
    (
        df
        .pivot_table(index=quantity, values=value, aggfunc='count')
        .plot(kind='pie', y=value, autopct='%1.1f%%', ylabel='', title=quantity, ax=ax[0]).legend(bbox_to_anchor=(1.9, 0.))
    );
    
    df_to_draw =(
        df
        .pivot_table(index=quantity, values=value, aggfunc='count').reset_index()
    );
    ax = sns.barplot(data=df_to_draw, x=quantity, y=value)
      
    ax.set(ylabel='Количество событий')
    fig.tight_layout()
```

```python
def dataframe_analysis(df):
    fig, ax=plt.subplots(nrows=1, ncols=1, figsize=(15, 15))
    print(df.describe())
    sns.boxplot(data=df, ax=ax)
    ax.set(ylabel='Количество событий')
    num_col_names = df.select_dtypes(include='number').columns.tolist()
    cat_col_names = df.select_dtypes(exclude='number').columns.tolist()
    for i in num_col_names:
        draw_quantitative(df, i)
    for i in cat_col_names:
        draw_categorize(df, i, num_col_names[0])
        
```

```python
#Подготовим пайплайн для обработки данных с учётом замены пустых строк на значение NAN

# создаём списки с названиями признаков
ohe_columns = ['workload', 'dept'] 
ord_columns = ['level', 'last_year_promo', 'last_year_violations', 'supervisor_evaluation']
num_columns = ['salary','employment_years']#' 
ohe_pipe = Pipeline(
    [('simpleImputer_ohe_1', SimpleImputer(missing_values=np.nan, strategy='most_frequent')),
     ('simpleImputer_ohe_2', SimpleImputer(missing_values=' ', strategy='most_frequent')),
     ('ohe', OneHotEncoder(drop='first', sparse_output=False, handle_unknown='ignore'))
    ]
    )

# создаём пайплайн для подготовки признаков из списка ord_columns: заполнение пропусков и Ordinal-кодирование
# SimpleImputer + OE
ord_pipe = Pipeline(
    [('simpleImputer_before_ord_1', SimpleImputer(missing_values=np.nan, strategy='most_frequent')),
     ('simpleImputer_before_ord_2', SimpleImputer(missing_values=' ', strategy='most_frequent')),
     ('ord',  OrdinalEncoder(
                categories=[
                ['junior', 'middle', 'sinior'],
                ['no', 'yes'],
                ['no', 'yes'],
                [1, 2, 3, 4, 5]],
                handle_unknown='use_encoded_value', unknown_value=np.nan
            )
        ),
     ('simpleImputer_after_ord', SimpleImputer(missing_values=np.nan, strategy='most_frequent'))
    ]
)

# создаём общий пайплайн для подготовки данных
data_preprocessor = ColumnTransformer(
    [('ohe', ohe_pipe, ohe_columns),
     ('ord', ord_pipe, ord_columns),
     ('num', MinMaxScaler(), num_columns),
    ], 
    remainder='passthrough'
)
```

## Шаг пятый

```python
columns_total = ohe_columns + ord_columns + num_columns
```

```python
# создаём пользовательскую метрику
def SMAPE_metric(y_true, y_pred):
    list_ = []
    for i in range(0, len(y_true)):
        list_.append(np.abs(y_true.iloc[i] - y_pred[i])/(y_true.iloc[i] + y_pred[i]))
    metric = sum(list_)* 100 * 2 / len(y_true)
    return metric

scorer = make_scorer(SMAPE_metric, greater_is_better=False) 
```

```python
RANDOM_STATE = 42

# создаём итоговый пайплайн: подготовка данных и модель
pipe_final = Pipeline([
    ('preprocessor', data_preprocessor),
    ('models', DecisionTreeRegressor())
 
])

param_grid = [
    # словарь для модели DecisionTreeClassifier()
    {
        'models': [DecisionTreeRegressor(random_state=RANDOM_STATE)],
        'models__max_depth': range(115, 130),
        'models__min_samples_split': range(1, 35),
        'models__criterion': ['squared_error', 'friedman_mse', 'absolute_error', 'poisson'],
        'models__splitter': ['best', 'random'],
        'preprocessor__num': [StandardScaler(), MinMaxScaler()],
    },
    # словарь для модели LogisticRegression()
    {
        'models': [LinearRegression()],
        'preprocessor__num': [StandardScaler(), MinMaxScaler()],    
    },
]
```

```python
# Обучим первую модель на DecisionTreeRegressor
model_Search1 = RandomizedSearchCV(
    pipe_final, 
    param_grid, 
    cv=5,
    n_iter=25,
    scoring=scorer,
    random_state=RANDOM_STATE,
    n_jobs=-1
)

model_Search1.fit(df_train[columns_total], df_train['job_satisfaction_rate']);

ohe_columns_transformed = model_Search1.best_estimator_['preprocessor'].named_transformers_['ohe']['ohe'].get_feature_names_out()
list_of_columns = []
for i in ohe_columns_transformed.tolist():
    list_of_columns.append(i)
for i in ord_columns:
    list_of_columns.append(i)
for i in num_columns:
    list_of_columns.append(i)


X_valid_tranformed = model_Search1.best_estimator_['preprocessor'].transform(df_train[columns_total])
X_valid_tranformed = pd.DataFrame(X_valid_tranformed, columns = list_of_columns)

X_test_tranformed = model_Search1.best_estimator_['preprocessor'].transform(df_test[columns_total])
X_test_tranformed = pd.DataFrame(X_test_tranformed, columns = list_of_columns)
    
print('Лучшая модель и её параметры:\n\n', model_Search1.best_estimator_)
print ('Метрика лучшей модели на кросс-валидационной выборке:', model_Search1.best_score_)

print(f'Метрика на тренировочной выборке: {scorer(model_Search1.best_estimator_["models"], X_valid_tranformed, df_train["job_satisfaction_rate"])}')
print(f'Метрика на тестовой выборке: {scorer(model_Search1.best_estimator_["models"], X_test_tranformed, df_test["job_satisfaction_rate"])}')
```

Главным его преимуществом является способность использовать различные подмножества признаков и правила принятия решений на разных этапах классификации, что позволяет делать классификацию жадным поиском методом "разделяй и властвуй".
Метрикой качества дял данной задачи являлась roc_auc, чьё значение на тестовой выборке составляет 0.92.

На основе полученных результатов можно порекомендовать бизнесу, что чтобы сотрудник не ушёл, надо пристально следить за его уровнем удовлетворённости работой, на которую влияет зарплата сотрудника, оценка его деятельности руководителем, рабочая наргузка и его уровень. То есть, для таких сотрудников можно улучшить взаимодействие руководителя с ним (давать поощрения за работу, примии, это может мотивировать сотрудника и как итог улучшится его оценка самим руководителем), повысить заплату (чтобы нагрузка на работника была сопоставима с его зарплатой), нужно повышение сотрудника (если долго он будет находится на уровне junior, то он решит, что карьерного роста у него нет и надо искать другую работу).
